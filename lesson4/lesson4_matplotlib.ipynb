```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
days_count = 100
data = {
    'day': np.arange(1, days_count + 1),
    "coffee_type": np.random.choice(
        ['Espresso', 'Latte', 'Cappuccino', 'Americano', 'Mocha'], days_count
    ),
    'cups_sold': np.random.randint(20,150, days_count),
    'temperature': np.random.uniform(15, 30, days_count),
    "customer_rating": np.random.uniform(2, 5, days_count)
}
df = pd.DataFrame(data)
print(df.head())
```

# Matlib 
Это библиотека, которая используется для создания графиков. Есть интергация с библиотеками Pandas та Numpy
1. Основные обьекты:
    * Фигура(Figure) - это холст или окно, на котором находится график.
    * Оси(Axes) - это график с координатаной сеткой, осями Х та Y . На одной фигуре может быть несколько графиков.
    * Графики(Plots) - это линии, точки или столбцы которые отображают данные.


```python
(fix, ax) = plt.subplots(nrows=2, ncols=3,figsize=(8,4))
ax[0,0].plot(
    df['day'], df['cups_sold'], color='blue'
)
ax[0,0].set_title('Cups Sold Over Time')
ax[0,0].set_xlabel('Day')
ax[0,0].set_ylabel('Cups Sold')
plt.show()
```

```python
# plt.figure(figsize=(8,4))
ax[0,1].plot(df['day'], df['temperature'], color='red', linestyle='--')
ax[0,1].set_title('Temperature Over Time')
ax[0,1].set_xlabel('Day')
ax[0,1].set_ylabel('Temperature (°C)')
ax[0,1].grid(True)
# plt.show()
```

```python
# plt.figure(figsize=(8,4))
ax[0,2].hist(df['temperature'], bins=10, color='skyblue', edgecolor='black')
ax[0,2].set_title('Temperature Distribution')
ax[0,2].set_xlabel('Temperature (°C)')
ax[0,2].set_ylabel('Frequency')
# plt.show()
```

```python
coffee_cups = df['coffee_type'].value_counts()
coffee_cups
```

```python
plt.figure(figsize=(8,4))
plt.pie(
    coffee_cups,
    labels=coffee_cups.index,
    autopct='%1.1f%%',
    colors=['green', 'blue', 'red','yellow', 'orange'],

)
```

```python
# plt.figure(figsize=(8,4))
plt.scatter(df['temperature'], df['cups_sold'], color='green')
plt.title('Cups Sold vs Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Cups Sold')
plt.grid(True)
# plt.show()
```

```python
avg_cups_by_coffee = df.groupby('coffee_type')['cups_sold'].mean()
plt.figure(figsize=(8, 4))
avg_cups_by_coffee.plot(
    kind='bar', 
    color='coral', 
    edgecolor='black'
)

plt.title('Середня кількість проданих чашок за типом кави')
plt.xlabel('Тип кави')
plt.ylabel('Середні продажі (чашок)')
plt.xticks(rotation=0)
plt.show()
```

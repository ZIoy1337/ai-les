## Cтатистика 

1. Випадкова величина це показник значення якого заздалегідь невідомо і залежить від випадку
2. Середнє (Mean) - сума всіх значень в діапазоні якому ділять на їх кількість. [1500, 1000, 1200, 1500, 1200, 5000] = 1900
3. Мода - найпопулярніше число зі списку
4. Медіана - це число, яке знаходиться рівно посередині відсортованого за зростанням або спаданням списку
5. Дисперсія і середньо-квадратичне відхилення показує наскільки сильно дані відрізняються навколо середнього значення
Дисперсія: 10

5 студентів, кожен з них викноує завдання. Дано діапазон чисел, скільки годин витратили на завдання. [2,4,6,8,18]
d SUM(xi середнє) / n-1
- знаходимо середнє:
4 + 6 + 8 + 10 = 30 / 5 = 6
- рахуємо відхилення кожного значення від середнього:
    - (2 - 6) ^ 2 = (- 4) ^ 2 = 16
    - (4 - 6) ^ 2 = (- 2) ^ 2 = 4
    - (6 - 6) ^ 2 = (0) ^ 2 = 0
    - (8 - 6) ^ 2 = (2) ^ 2 = 4
    - (10 - 6) ^ 2 = (4) ^ 2 = 16
- Сумуємо квадрати: 16 + 4 + 0 + 4 + 16 = 40
40 / 4 = 10
Дисперсія: 10
sqrt(10) = 3.16 годин
 

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
students_count = 500
```

```python
data = {
    'hours_studied': np.clip(np.random.normal(loc = 15, scale =4, size=students_count), 2, 30),
    'projects_completed': np.random.poisson(lam=3, size=students_count),
    'exam_score': np.clip(np.random.normal(loc = 75, scale =10, size=students_count), 0, 100)
}
data
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
print(f'''
    average hours studied: {df['hours_studied'].mean()}
    average exam score : {df['exam_score'].mean()}
    meadian hours studied: {df['hours_studied'].median()}
      ''')
```

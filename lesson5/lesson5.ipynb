```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
n_samples = 100

experience = np.random.uniform(0, 15, n_samples)
salary = 15000 + experience * 3000 + np.random.normal(0, 5000, n_samples)

plt.figure(figsize=(8, 5))
plt.scatter(experience, salary, color='blue', alpha=0.7, edgecolor='black')

plt.title('Залежність зарплати від стажу роботи')
plt.xlabel('Стаж роботи (років)')
plt.ylabel('Зарплата (грн)' )
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

# task 2

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n_samples = 120

revenue = np.random.uniform(50, 1000, n_samples)
saturation_limit = 150
marketing_spend = saturation_limit * (1 - np.exp(-0.0045 * revenue)) + np.random.normal(0, 5, n_samples)
marketing_spend = np.clip(marketing_spend, 5, None)

x_curve = np.linspace(50, 1000, 300)
y_curve = saturation_limit * (1 - np.exp(-0.0045 * x_curve))

plt.figure(figsize=(9, 5))
plt.scatter(revenue, marketing_spend, color='teal', alpha=0.7, edgecolor='black', label='Дані компанії')
plt.plot(x_curve, y_curve, color='red', linewidth=2, label='Крива насичення')
plt.axhline(saturation_limit, color='gray', linestyle='--', label='Поріг насичення ринку')

plt.title('Залежність витрат на маркетинг від виручки (з ефектом насичення)', fontsize=13)
plt.xlabel('Виручка компанії (тис. грн)', fontsize=11)
plt.ylabel('Витрати на маркетинг (тис. грн)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

# task 3

```python
np.random.seed(42)
n_samples = 150

hours_worked = np.random.uniform(30, 60, n_samples)
urgency = np.random.uniform(1, 10, n_samples)

stress_level = 1.2 * hours_worked + 4.5 * urgency + np.random.normal(0, 5, n_samples)
stress_level = np.clip(stress_level, 1, 100)

plt.figure(figsize=(8, 5))
sc1 = plt.scatter(hours_worked, urgency, c=stress_level, cmap='coolwarm', edgecolor='black', s=50)
plt.colorbar(sc1, label='Stress Level')
plt.title('Stress Level vs Hours Worked and Urgency')
plt.xlabel('Hours Worked Per Week')
plt.ylabel('Urgency (1-10)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
price = np.random.uniform(50, 500, n_samples)
quantity = 5000 / price + np.random.normal(0, 2, n_samples)
quantity = np.clip(quantity, 1, None)

plt.figure(figsize=(8, 5))
plt.scatter(price, quantity, color='purple', alpha=0.7, edgecolor='black')
plt.title('Relationship Between Product Price and Purchased Quantity')
plt.xlabel('Product Price (UAH)')
plt.ylabel('Quantity Purchased')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

# task 5

```python
hours_of_day = np.random.uniform(0, 24, n_samples)
weather_conditions = np.random.choice(['Clear', 'Rain', 'Snow'], size=n_samples, p=[0.6, 0.3, 0.1])

base_travel_time = 30
travel_time = []

for h, w in zip(hours_of_day, weather_conditions):
    t = base_travel_time
    if (7 <= h <= 9) or (17 <= h <= 19):
        t += 25
    if w == 'Rain':
        t += 15
    elif w == 'Snow':
        t += 35
    t += np.random.normal(0, 5)
    travel_time.append(max(10, t))

travel_time = np.array(travel_time)

plt.figure(figsize=(8, 5))
for weather, color in zip(['Clear', 'Rain', 'Snow'], ['orange', 'blue', 'gray']):
    mask = weather_conditions == weather
    plt.scatter(hours_of_day[mask], travel_time[mask], color=color, label=weather, alpha=0.8, edgecolor='black')

plt.title('Travel Time Simulation by Time of Day and Weather')
plt.xlabel('Hour of Day')
plt.ylabel('Travel Time (min)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

# task 6

```python
wait_time = np.random.uniform(5, 60, n_samples)
service_quality = np.random.uniform(1, 10, n_samples)

rating = 10 * (service_quality / 10) * (1 - wait_time / 70) + np.random.normal(0, 0.3, n_samples)
rating = np.clip(rating, 1, 5)

plt.figure(figsize=(8, 5))
sc4 = plt.scatter(wait_time, service_quality, c=rating, cmap='viridis', edgecolor='black', s=50)
plt.colorbar(sc4, label='Final Restaurant Rating (1-5)')
plt.title('Restaurant Rating vs Wait Time and Service Quality')
plt.xlabel('Order Wait Time (min)')
plt.ylabel('Service Quality (1-10)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
